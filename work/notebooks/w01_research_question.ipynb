```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

visible = df[df["impressions_90d"] >= 100]
group_counts = visible.groupby(["position_tier", "content_type"])["ctr"].count()
print("Sample sizes by position_tier x content_type:")
print(group_counts)

keyword_ctr = visible[visible["content_type"] == "keyword_article"].groupby("position_tier")["ctr"].mean().round(4)
print("\nCTR for keyword_article by position tier:")
print(keyword_ctr)
```

```python
keyword_ctr = visible[visible["content_type"] == "keyword article"].groupby("position_tier")["ctr"].mean().round(4)
print("\nCTR for keyword article by position tier:")
print(keyword_ctr)
```

```python
## Real numbers from the starter dataset:
# Sample sizes vary enormously by content_type and position_tier. keyword article has 527-8,186 rows per tier, while comparison article and feedly article have as
# few as 1 and 5 rows. This confirms that any CTR opportunity score must weight by sample confidence, not just raw CTR, otherwise the "best" and "worst" pages found will just be sampling noise.

# CTR for keyword article follows a clear, trustworthy pattern by position:0.3104 at top_3, 0.3458 at page_1, dropping to 0.1427 at page_3_5, 0.2559 at
#striking, and 0.0555 at deep.

# Some raw CTR values exceed 1.0 (e.g. 2.90 for feedly article) which is mathematically impossible.

```
